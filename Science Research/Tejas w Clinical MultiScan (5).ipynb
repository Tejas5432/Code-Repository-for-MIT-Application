```python
import os
import glob
import numpy as np
import pandas as pd
import pydicom
import cv2
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
from sklearn.metrics import confusion_matrix
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import VGG16
from tensorflow.keras.models import Model

IMG_SIZE = 240
NUM_SLICES = 70

def load_dicom_series(series_path, target_size=(IMG_SIZE, IMG_SIZE), target_depth=NUM_SLICES):
    dicom_files = glob.glob(os.path.join(series_path, '*.dcm'))
    if len(dicom_files) == 0:
        return None
    dicom_files = sorted(dicom_files)
    slices = []
    for dcm_file in dicom_files:
        try:
            ds = pydicom.dcmread(dcm_file)
            img = ds.pixel_array.astype(np.float32)
            img = (img - np.min(img)) / (np.max(img) - np.min(img) + 1e-6)
            img = cv2.resize(img, target_size)
            slices.append(img)
        except Exception:
            continue
    if len(slices) == 0:
        return None
    slices = np.array(slices)
    current_depth = slices.shape[0]
    if current_depth != target_depth:
        indices = np.linspace(0, current_depth - 1, target_depth).astype(np.int32)
        slices = slices[indices, :, :]
    slices = np.expand_dims(slices, axis=-1)
    return slices

clinical_df = pd.read_csv('Complete.csv')
clinical_df = clinical_df[clinical_df['Grade'].notnull()]
clinical_df = clinical_df[clinical_df['Grade'] != 'Empty']

grade_mapping = {
    'Well Differentiated (G1)': 0,
    'Moderately Differentiated (G2)': 1,
    'Poorly Differentiated (G3)': 2,
    'Undifferentiated (G4)': 3
}
clinical_df['GradeClass'] = clinical_df['Grade'].map(grade_mapping)

categorical_features = ['race', 'cigsmok', 'gender']
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encoded_cats = encoder.fit_transform(clinical_df[categorical_features])

scaler = MinMaxScaler()
clinical_df['age_scaled'] = scaler.fit_transform(clinical_df[['age']])

clinical_features = np.hstack([encoded_cats, clinical_df[['age_scaled']].values])

X_img, X_clinical, y = [], [], []
excluded_patients = 0

patient_dirs = [d for d in os.listdir('.') if os.path.isdir(d) and d.isdigit()]
for patient_dir in patient_dirs:
    pid = patient_dir
    record = clinical_df[clinical_df['pid'].astype(str) == pid]
    if record.empty:
        continue
    label = record['GradeClass'].values[0]
    clinical_vector = clinical_features[clinical_df.index[clinical_df['pid'].astype(str) == pid].tolist()[0]]

    patient_path = os.path.join('.', patient_dir)
    scan_sessions = [d for d in os.listdir(patient_path) if os.path.isdir(os.path.join(patient_path, d))]
    if len(scan_sessions) < 2:
        excluded_patients += 1
        continue
    sessions = sorted(scan_sessions)
    volumes = []

    for sess in sessions[:2]:
        session_path = os.path.join(patient_path, sess)
        series_folders = [d for d in os.listdir(session_path) 
                          if os.path.isdir(os.path.join(session_path, d)) and d.startswith('2')]
        if not series_folders:
            series_folders = [d for d in os.listdir(session_path) 
                              if os.path.isdir(os.path.join(session_path, d)) and d.startswith('3')]
        if not series_folders:
            continue
        series_path = os.path.join(session_path, series_folders[0])
        volume = load_dicom_series(series_path)
        if volume is None:
            continue
        volumes.append(volume)
    
    if len(volumes) == 2:
        X_img.append(np.stack(volumes, axis=0))
        X_clinical.append(clinical_vector)
        y.append(label)

X_img = np.array(X_img)
X_clinical = np.array(X_clinical)
y = np.array(y)

X_img_train, X_img_test, X_clinical_train, X_clinical_test, y_train, y_test = train_test_split(
    X_img, X_clinical, y, test_size=0.2, random_state=42
)

vol_input = layers.Input(shape=(2, NUM_SLICES, IMG_SIZE, IMG_SIZE, 1))
x = layers.TimeDistributed(layers.Reshape((NUM_SLICES, IMG_SIZE, IMG_SIZE)))(vol_input)
x = layers.TimeDistributed(layers.Lambda(lambda img: tf.image.grayscale_to_rgb(tf.expand_dims(img, axis=-1))))(x)
x = layers.Reshape((2 * NUM_SLICES, IMG_SIZE, IMG_SIZE, 3))(x)

vgg_base = VGG16(include_top=False, weights='imagenet', input_shape=(IMG_SIZE, IMG_SIZE, 3))
for layer in vgg_base.layers[:-4]:
    layer.trainable = False

x = layers.TimeDistributed(vgg_base)(x)
x = layers.TimeDistributed(layers.GlobalAveragePooling2D())(x)
x = layers.Reshape((2, NUM_SLICES, -1))(x)
x = layers.GlobalAveragePooling2D()(x)  # Reduce both the timepoint and slice dimensions
x = layers.Dense(256, activation='relu')(x)
x = layers.Dropout(0.5)(x)

clin_input = layers.Input(shape=(X_clinical.shape[1],))
y = layers.Dense(64, activation='relu')(clin_input)
y = layers.Dense(32, activation='relu')(y)

merged = layers.concatenate([x, y])
z = layers.Dense(256, activation='relu')(merged)
z = layers.Dropout(0.5)(z)
output = layers.Dense(4, activation='softmax')(z)

model = Model(inputs=[vol_input, clin_input], outputs=output)
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()

history = model.fit(
    [X_img_train, X_clinical_train], y_train,
    epochs=20, batch_size=2, validation_data=([X_img_test, X_clinical_test], y_test)
)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='train_loss')
plt.plot(history.history['val_loss'], label='val_loss')
plt.legend()
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='train_acc')
plt.plot(history.history['val_accuracy'], label='val_acc')
plt.legend()
plt.show()
```
